```python
import pandas as pd
import numpy as np

# Load cleaned dataset
df = pd.read_excel("Cleaned_Application_Funnel.xlsx")

# Basic information
total_applications = len(df)
total_students = df["Student_ID"].nunique()
total_companies = df["Company"].nunique()

# Application status counts
status_summary = (
    df["Application_Status"]
    .value_counts()
    .reset_index()
)

status_summary.columns = [
    "Application_Status",
    "Applications"
]

status_summary["Percentage"] = (
    status_summary["Applications"] / total_applications * 100
).round(2)

# Funnel metrics
applied = (df["Application_Status"] == "Applied").sum()
shortlisted = (df["Application_Status"] == "Shortlisted").sum()
interview = (df["Application_Status"] == "Interview").sum()
selected = (df["Application_Status"] == "Selected").sum()
rejected = (df["Application_Status"] == "Rejected").sum()

funnel_summary = pd.DataFrame({
    "Metric": [
        "Total Applications",
        "Applied",
        "Shortlisted",
        "Interview",
        "Selected",
        "Rejected"
    ],
    "Count": [
        total_applications,
        applied,
        shortlisted,
        interview,
        selected,
        rejected
    ]
})

# Conversion metrics
shortlist_rate = round(
    shortlisted / total_applications * 100, 2
)

interview_rate = round(
    interview / total_applications * 100, 2
)

selection_rate = round(
    selected / total_applications * 100, 2
)

rejection_rate = round(
    rejected / total_applications * 100, 2
)

conversion_summary = pd.DataFrame({
    "Metric": [
        "Shortlist Rate",
        "Interview Rate",
        "Selection Rate",
        "Rejection Rate"
    ],
    "Percentage": [
        shortlist_rate,
        interview_rate,
        selection_rate,
        rejection_rate
    ]
})

# Company-wise analysis
company_summary = (
    df.groupby("Company")
    .agg(
        Applications=("Application_ID", "count"),
        Students=("Student_ID", "nunique"),
        Average_Processing_Days=("Processing_Days", "mean")
    )
    .reset_index()
)

company_summary["Average_Processing_Days"] = (
    company_summary["Average_Processing_Days"].round(2)
)

# Company status analysis
company_status = pd.crosstab(
    df["Company"],
    df["Application_Status"]
).reset_index()

# Skill-wise analysis
skill_summary = (
    df.groupby("Skill")
    .agg(
        Applications=("Application_ID", "count"),
        Students=("Student_ID", "nunique"),
        Average_Experience=("Experience_Years", "mean"),
        Average_Processing_Days=("Processing_Days", "mean")
    )
    .reset_index()
)

skill_summary["Average_Experience"] = (
    skill_summary["Average_Experience"].round(2)
)

skill_summary["Average_Processing_Days"] = (
    skill_summary["Average_Processing_Days"].round(2)
)

# Education-wise analysis
education_summary = (
    df.groupby("Education")
    .agg(
        Applications=("Application_ID", "count"),
        Students=("Student_ID", "nunique"),
        Average_Experience=("Experience_Years", "mean")
    )
    .reset_index()
)

education_summary["Average_Experience"] = (
    education_summary["Average_Experience"].round(2)
)

# Region-wise analysis
region_summary = (
    df.groupby("Region")
    .agg(
        Applications=("Application_ID", "count"),
        Students=("Student_ID", "nunique")
    )
    .reset_index()
)

# Gender-wise analysis
gender_summary = (
    df.groupby("Gender")
    .agg(
        Applications=("Application_ID", "count"),
        Students=("Student_ID", "nunique")
    )
    .reset_index()
)

# Experience-wise analysis
experience_summary = (
    df.groupby("Experience_Years")
    .agg(
        Applications=("Application_ID", "count"),
        Average_Processing_Days=("Processing_Days", "mean")
    )
    .reset_index()
)

experience_summary["Average_Processing_Days"] = (
    experience_summary["Average_Processing_Days"].round(2)
)

# Monthly application trend
df["Month"] = df["Application_Date"].dt.month
df["Month_Name"] = df["Application_Date"].dt.strftime("%B")

monthly_summary = (
    df.groupby(["Month", "Month_Name"])
    .agg(
        Applications=("Application_ID", "count"),
        Students=("Student_ID", "nunique")
    )
    .reset_index()
    .sort_values("Month")
)

# Processing time analysis
processing_summary = pd.DataFrame({
    "Metric": [
        "Average Processing Days",
        "Minimum Processing Days",
        "Maximum Processing Days",
        "Median Processing Days"
    ],
    "Value": [
        round(df["Processing_Days"].mean(), 2),
        df["Processing_Days"].min(),
        df["Processing_Days"].max(),
        df["Processing_Days"].median()
    ]
})

# Overall KPI summary
kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Applications",
        "Unique Students",
        "Companies",
        "Shortlisted",
        "Interview",
        "Selected",
        "Rejected",
        "Shortlist Rate",
        "Interview Rate",
        "Selection Rate",
        "Average Processing Days"
    ],
    "Value": [
        total_applications,
        total_students,
        total_companies,
        shortlisted,
        interview,
        selected,
        rejected,
        shortlist_rate,
        interview_rate,
        selection_rate,
        round(df["Processing_Days"].mean(), 2)
    ]
})

# Data quality summary
data_quality = pd.DataFrame({
    "Check": [
        "Total Rows",
        "Total Columns",
        "Duplicate Rows",
        "Missing Values",
        "Unique Applications",
        "Unique Students",
        "Unique Companies"
    ],
    "Value": [
        df.shape[0],
        df.shape[1],
        df.duplicated().sum(),
        df.isnull().sum().sum(),
        df["Application_ID"].nunique(),
        df["Student_ID"].nunique(),
        df["Company"].nunique()
    ]
})

# Business insights
top_company = (
    company_summary
    .sort_values("Applications", ascending=False)
    .iloc[0]["Company"]
)

top_skill = (
    skill_summary
    .sort_values("Applications", ascending=False)
    .iloc[0]["Skill"]
)

top_region = (
    region_summary
    .sort_values("Applications", ascending=False)
    .iloc[0]["Region"]
)

top_education = (
    education_summary
    .sort_values("Applications", ascending=False)
    .iloc[0]["Education"]
)

business_insights = pd.DataFrame({
    "Insight": [
        "Most active company",
        "Most applied skill",
        "Region with highest applications",
        "Education group with highest applications",
        "Overall shortlist rate",
        "Overall interview rate",
        "Overall selection rate",
        "Average processing time"
    ],
    "Result": [
        top_company,
        top_skill,
        top_region,
        top_education,
        f"{shortlist_rate}%",
        f"{interview_rate}%",
        f"{selection_rate}%",
        f"{round(df['Processing_Days'].mean(), 2)} days"
    ]
})

# Save all analysis results into ONE Excel file
with pd.ExcelWriter(
    "Python Analysis Result.xlsx",
    engine="openpyxl"
) as writer:

    kpi_summary.to_excel(
        writer,
        sheet_name="KPI Summary",
        index=False
    )

    funnel_summary.to_excel(
        writer,
        sheet_name="Application Funnel",
        index=False
    )

    conversion_summary.to_excel(
        writer,
        sheet_name="Conversion Rates",
        index=False
    )

    status_summary.to_excel(
        writer,
        sheet_name="Status Analysis",
        index=False
    )

    company_summary.to_excel(
        writer,
        sheet_name="Company Analysis",
        index=False
    )

    company_status.to_excel(
        writer,
        sheet_name="Company Status",
        index=False
    )

    skill_summary.to_excel(
        writer,
        sheet_name="Skill Analysis",
        index=False
    )

    education_summary.to_excel(
        writer,
        sheet_name="Education Analysis",
        index=False
    )

    region_summary.to_excel(
        writer,
        sheet_name="Region Analysis",
        index=False
    )

    gender_summary.to_excel(
        writer,
        sheet_name="Gender Analysis",
        index=False
    )

    experience_summary.to_excel(
        writer,
        sheet_name="Experience Analysis",
        index=False
    )

    monthly_summary.to_excel(
        writer,
        sheet_name="Monthly Analysis",
        index=False
    )

    processing_summary.to_excel(
        writer,
        sheet_name="Processing Analysis",
        index=False
    )

    business_insights.to_excel(
        writer,
        sheet_name="Business Insights",
        index=False
    )

    data_quality.to_excel(
        writer,
        sheet_name="Data Quality",
        index=False
    )

print("Analysis completed successfully.")
print("Output file: Python Analysis Result.xlsx")
print("Total Applications:", total_applications)
print("Unique Students:", total_students)
print("Companies:", total_companies)
print("Shortlist Rate:", shortlist_rate, "%")
print("Interview Rate:", interview_rate, "%")
print("Selection Rate:", selection_rate, "%")
print("Average Processing Days:", round(df["Processing_Days"].mean(), 2))
```
